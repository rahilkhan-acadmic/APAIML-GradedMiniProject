# Operational Analysis

    ## Retrieve the top 10 departments by total visit volume.


```python
select department, count(department) as total_visits from visits group by department order by total_visits desc limit 10;
```

## Identify the top 5 departments with the highest average length of stay.

```python
select department, avg(length_of_stay_hours) as avg_length_of_stay_hours
from visits
group by department
order by avg_length_of_stay_hours desc limit 5;
```

## Find the percentage of High Risk visits per department.

```python

select department,
count(*) as total_visits,
sum(case when risk_score ='High' Then 1 else 0 END) as high_risk_visits,
round( sum(case when risk_score='High' then 1 else 0 end) * 100 / count(*), 2) as high_risk_percentage
from visits
group by department;
```

##  Determine the average number of visits per patient by city.

```python

select
p.city,
Round(count(v.visit_id) * 1.0/count(distinct p.patient_id),2) as avg_visits_per_patient
from visits v
join patients p on v.patient_id = p.patient_id
group by p.city;
```

## Identify doctors handling the highest number of High Risk visits.
### Note: No. of doctors is not defined as such top 10 doctors with highest no. of high risk visits are fetched.


```python
select count(risk_score) as high_risk_count, doctor_id
from visits where risk_score = 'High'
group by  doctor_id
order by high_risk_count desc Limit 10;
```
