# Model Evaluation & Documentation: Hospital Operations & Finance

**Business Goal:** Ensure that predictions are reliable, interpretable, and safe for hospital operations and finance teams.

---

## 1. Risk Model Evaluation Report (Hospital Visit Risk)
*Classifies hospital visits as Low, Medium, or High risk.*

### Performance Summary

| Model | Overall Accuracy | High Risk F1-Score |
| :--- | :--- | :--- |
| **Logistic Regression (Baseline)** | ~33% | 0.26 |
| **Random Forest (Tuned)** | **~46%** | **~0.10** |

### Key Business Metrics
*   **Recall for High Risk Class:** **~0.10**.
    *   *Analysis:* The model captures only 10% of actual high-risk visits. This poses a safety risk for operations if used for automated staffing, as 90% of high-risk cases are missed.
*   **Confusion Matrix Insight:** The model frequently misclassifies "High" and "Medium" risk visits as "Low" risk due to the overwhelming historical volume of low-risk data.

### Fairness & Segmentation
*   **By Gender/City:** Performance is consistent across demographics, indicating no significant algorithmic bias.
*   **By Chronic Flag:** Accuracy is slightly higher for patients with a `chronic_flag`, as their visit patterns are more predictable.

## 2. Claim Model Evaluation Report (Insurance Claim Outcome)
*Predicts if a claim will be Paid, Pending, or Rejected.*

### Performance Summary

| Model | Overall Accuracy | Paid F1-Score | Pending F1-Score | Rejected F1-Score |
| :--- | :--- | :--- | :--- | :--- |
| **Logistic Regression** | 46% | 0.63 | 0.18 | 0.28 |
| **Random Forest** | **56%** | **0.70** | **0.23** | **0.25** |

### Key Business Metrics
*   **Recall for Rejected Claim Class:** **0.21** (Random Forest).
    *   *Analysis:* The finance team can only pre-identify 21% of claims that will eventually be rejected. 79% of rejections remain "surprises," limiting the model's current utility for cash-flow forecasting.

### Fairness & Segmentation
*   **By Insurance Provider:** The model is most accurate for "SecureLife" (larger data share) and least accurate for "MediCareX" regarding rejections.
*   **Fairness Gap:** There is a 12% accuracy gap between the top-performing city (Hyderabad) and the lowest (Pune), likely due to regional billing variations.

## 3. Model Card Document

### Model Details
*   **Type:** Random Forest Classifier (Ensemble Learning).
*   **Target Variables:** `risk_score` and `claim_status`.
*   **Training Data:** 20,000 records; 80/20 time-based split to prevent data leakage.

### Intended Use
*   **Risk Model:** Advisory tool for clinical resource allocation.
*   **Claim Model:** Pre-submission check for finance teams to flag potential billing errors.

### Assumptions & Limitations
*   **Assumption:** Future billing cycles and clinical protocols remain consistent with the 2025 training data.
*   **Limitation:** Despite using **SMOTE** to balance training classes, the model struggles with the inherent "noise" and overlap between Pending and Rejected claims.

## 4. Explainability Summary

The models prioritize features that represent **intensity** and **discrepancy**:

1.  **Operational Intensity:** `length_of_stay_days` and `department` (e.g., ICU) are the top predictors for Risk. Longer stays directly correlate to higher risk scores.
2.  **Financial Discrepancy:** `amount_difference` (Billed vs. Approved) is the primary driver for Claim Status. A high difference is a near-certain indicator of a "Rejected" or "Pending" status.
3.  **Patient Stability:** `avg_days_between_visits` serves as a proxy for patient health stability. Frequent visits often trigger higher risk flags.

**Conclusion:** Both models are currently **Under-performing** for minority classes (High Risk/Rejected). They should be used for **human-in-the-loop** decision support rather than automated processing.
